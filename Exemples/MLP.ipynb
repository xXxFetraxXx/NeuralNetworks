# MLP — Multi-Layer Perceptron

Le `MLP` sert à **approximer une fonction** : il apprend une correspondance entrées → sorties à partir de données.
Cas d'usage typiques : régression, reconstruction d'images à partir de coordonnées (x, y), classification de vecteurs.

Il est composé de couches entièrement connectées (`Linear` + activation), avec un **encodage de Fourier optionnel**
qui aide le réseau à apprendre des détails fins (utile pour les coordonnées spatiales).

```python
from NeuralNetworks import MLP
import torch
```

## Paramètres du constructeur

| Paramètre | Rôle |
|---|---|
| `input_size` | dimension de chaque entrée |
| `output_size` | dimension de chaque sortie |
| `hidden_layers` | tailles des couches cachées, ex. `[256, 256]` |
| `norm` | activation (`"Relu"`, `"Tanh"`, `"GELU"`...) |
| `fourier_input_size` | nb de dimensions d'entrée encodées en Fourier |
| `nb_fourier` | nb de fréquences de l'encodage |
| `sigmas` | échelles des fréquences (`None` = pas d'encodage) |
| `name` | nom du modèle, utilisé pour la sauvegarde |

Sans `sigmas`, l'encodage Fourier est désactivé : le MLP devient un réseau fully-connected classique.

```python
net = MLP(input_size=2, output_size=3, hidden_layers=[256] * 2, norm="Relu",
          fourier_input_size=2, nb_fourier=256, sigmas=[10], name="MLP")
```

## Forward

- entrée `[input_size]` → sortie `[output_size]`
- entrée `[batch, input_size]` → sortie `[batch, output_size]`

Le forward d'inférence se fait **sans gradients** et renvoie le résultat sur CPU.

```python
x = torch.rand(10, 2)   # batch de 10 entrées
y = net(x)
print(y.shape)          # (10, 3)
```

## Sauvegarde / chargement

`net.save` enregistre les poids **et** les hyperparamètres dans `<name>.npz`.
`MLP.load(path)` reconstruit un objet identique et recharge les poids.

```python
net.save                              # écrit MLP.npz
net2 = MLP.load("MLP.npz")
print(torch.allclose(net(x), net2(x)))   # True
```
